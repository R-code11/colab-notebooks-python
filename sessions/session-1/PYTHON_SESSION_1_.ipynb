# 7. Numbers

Python can work with numbers.

There are two basic number types we will use today.

### Integer — `int`

Whole numbers:

- `18`
- `20`
- `95`

### Float — `float`

Decimal numbers:

- `1.75`
- `32.5`
- `86.5`

We can use `type()` to check the type of a value.

````python
age = 18
height = 1.75

print("Age:", age)
print("Height:", height)

print("Type of age:", type(age))
print("Type of height:", type(height))
````

# 9. Boolean Values

A Boolean has only two possible values:

- `True`
- `False`

Booleans are useful when we want to represent something that is either true or false.

````python
is_student = True
has_python_experience = False

print("Is student:", is_student)
print("Has Python experience:", has_python_experience)

print("Type:", type(is_student))
````

# 10. Basic Arithmetic

Python can perform mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `**` | Power | `2 ** 3` |
| `%` | Remainder | `10 % 3` |

````python
print("Addition:", 10 + 5)
print("Subtraction:", 10 - 5)
print("Multiplication:", 10 * 5)
print("Division:", 10 / 5)
print("Power:", 2 ** 3)
print("Remainder:", 10 % 3)
````

# 11. Real Example — Marks Calculator

Let's calculate the total and average marks of a student.

We can store each subject's marks in a variable and then perform calculations.

````python
biology = 85
chemistry = 78
physics = 90

total = biology + chemistry + physics
mean = total / 3

print("Total marks:", total)
print("Average marks:", mean)
````

# 16.Interactive Marks Calculator

Now let's combine everything we have learned.

We will:

1. Ask for the student's name
2. Ask for marks
3. Calculate total marks
4. Calculate average marks
5. Display the result

````python


name = input("Enter your name: ")

biology = float(input("Enter Biology marks: "))
chemistry = float(input("Enter Chemistry marks: "))
physics = float(input("Enter Physics marks: "))

total = biology + chemistry + physics
average = total / 3

print("\n------------- RESULT -------------")
print("Student:", name)
print("Total Marks:", total)
print("Average Marks:", average)
print("----------------------------------")
````
