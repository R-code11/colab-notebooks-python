# 1. Conditional Statements

A conditional statement allows Python to make decisions.

A condition can have two possible results:

- `True`
- `False`

The basic structure is:

```python
if condition:
    statement

````python
#```python
temperature = 35

if temperature > 30:
    print("It is hot")
````

# 2. How `if` Works

The `if` statement asks Python to check a condition.

```python
temperature > 30

# 3. `if` and `else`

Sometimes we want Python to do something when a condition is `True`
and something different when it is `False`.

For this, we use `else`.

### Structure

```python
if condition:
    statement_1
else:
    statement_2

````python


#python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote yet")
````

# 4. `if`, `elif` and `else`

Sometimes there are more than two possible outcomes.

For example, a student's marks could give different grades.

We can use `elif` to check additional conditions.

### Structure

```python
if condition_1:
    statement_1
elif condition_2:
    statement_2
elif condition_3:
    statement_3
else:
    statement_4

````python


marks = 82

if marks >= 90:
  print("Grade A")
elif marks >= 55:
    print("Grade D")
elif marks >= 60:
    print("Grade C")
elif marks >= 70:
    print("Grade  B")
else:
    print("Grade F")
````

# 5. Comparison Operators

Comparison operators allow us to compare two values.

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |
| `==` | Equal to |
| `!=` | Not equal to |

A comparison produces either:

- `True`
- `False`

````python
x = 10

print(x > 5)
print(x < 5)
print(x >= 10)
print(x <= 8)
print(x == 10)
print(x != 7)
````

# 6. Logical Operators

Logical operators allow us to combine conditions.

Python has three important logical operators:

- `and`
- `or`
- `not`

## `and`

Both conditions must be `True`.

```text
True and True   → True
True and False  → False
False and True  → False
False and False → False

````python


age = int(input("enter your age:"))
has_id = True

# : (colon) → Starts a code block after statements like if, else, for, and while.

if age >= 18 and has_id:

    print("Entry allowed")
    print("Welcome")
else:
    print("Entry denied")

# and means both conditions must be true.

````

````python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("It is the weekend")
# or means at least one condition must be true.
````

````python
raining = 1

if not raining:
    print("You can go outside")

# This program checks whether it is raining or not.
# First, we assign False to the variable raining,
# which means it is not raining. Then, we use the
# not operator to reverse the value. Since raining
# is False, not raining becomes True. Therefore,
# the condition is satisfied, and the program prints You can go outside.
````

# 7. Combining Conditions

Comparison operators and logical operators can be used together.

For example, we can check whether a student's marks are between 60 and 100.

```python
marks >= 60 and marks <= 100

````python


marks = 75

if marks >= 60 and marks <= 100:
    print("Valid marks in the required range")
````

# 8. `for` Loops

A loop allows us to repeat a task.

A `for` loop is commonly used when:

- We want to go through a sequence
- We want to process items in a list
- We know approximately how many times something should happen

### Structure

```python
for variable in sequence:
    statement

````python

# code

# list
# sequence

# ek-ek karke
# in ka matlab hai ki hum kisi sequence se ek-ek karke values le rahe hain.
for number in range(1, 6):
    print(number)
````

# 11. Looping Through a List

A `for` loop can be used to process every item in a list.

For example:

```python
animals = ["cat", "dog", "rabbit"]

````python
#CODE

animals = ["cat", "dog", "rabbit", "elephant"]

for pets in animals:
    print(pets)
````

# 12. `for` Loop + `if`

Loops and conditions can be combined.

This allows us to:

1. Take one item
2. Check a condition
3. Perform an action if the condition is true
4. Move to the next item

This pattern is very useful when working with data.

````python
marks = [45, 78, 91, 56, 84]
for mark in marks:
  if mark >= 80:
        print(mark)
````
