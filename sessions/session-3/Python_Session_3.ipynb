# **While loop**


1. **A `while` loop runs as long as its condition is `True`.** As soon as the condition becomes `False`, the loop stops.
2. Think of it like this: *"Keep the tap open until the bucket is full."* Here, "bucket is not full" is the condition, and "keeping the tap open" is the task.
3. Every `while` loop must have **3 things**:
   - **Starting value**
   - **Condition**
   - **Update**


   ### The Bucket Analogy

   Imagine you are filling a bucket with water.
   *  **Starting Value (The Beginning):** You have an **empty bucket**. (0 litres)
   * **Condition (When to stop):** You want to fill it up to **10 litres**. So you keep checking: *"Is the bucket less than 10 litres?"*
   * **Update (The Action):** Every time, you pour **1 litre of water** into the bucket. This increases the water.
   * If you **forget the Update** (pouring water), the bucket will never fill up, and you'll keep checking forever — *"Is it full? No. Is it full? No."* **This is an Infinite Loop!**


````python
water = 0               # 1. Starting Value
while water < 10:       # 2. Condition
    print("Filling...", water, "litres")
    water = water + 1   # 3. Update
print("Bucket is full!")
````

**Example**
*   i = 1 → start counting from 1.
*   while i <= 5: → run the loop as long as i is less than or equal to 5.
* print(i) → print the current number.
* i = i + 1 → increase i by 1 every time.


````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
````

````python
while True:
  print("This will go on forever!")
# This loop runs forever because the condition is always True.
````

## Understanding `df.describe(include='all')`

When you use `include='all'`, Pandas combines the summary of **numbers** and **text**.

### Numbers vs. Text
Because numbers and text are different, Pandas shows **NaN** (Not a Number) where a statistic doesn't apply.

| Statistic | Applies to Numbers? | Applies to Text? | Meaning |
|---|---|---|---|
| `count` | ✅ | ✅ | How many values are filled |
| `unique` | ❌ | ✅ | How many different values exist |
| `top` | ❌ | ✅ | The most common value |
| `freq` | ❌ | ✅ | How many times the `top` value appears |
| `mean` | ✅ | ❌ | Average value |
| `std` | ✅ | ❌ | How spread out the data is |
| `min` | ✅ | ❌ | Smallest value |
| `25%` | ✅ | ❌ | 25th percentile (Q1) |
| `50%` | ✅ | ❌ | Median (middle value) |
| `75%` | ✅ | ❌ | 75th percentile (Q3) |
| `max` | ✅ | ❌ | Largest value |

### How to Read Your Specific Table

**1. For Numeric Columns**
- Look at the **`mean`**, **`min`**, **`50%`** (median), and **`max`** rows.
- Example: The average `marks` is **69.17**, and the highest is **100**.
- Ignore `unique`, `top`, `freq` (they show `NaN`).

**2. For Text Columns**
- Look at the **`unique`**, **`top`**, and **`freq`** rows.
- Example: The most common `grade` is **D** (appearing **26** times).
- Ignore `mean`, `std`, `min`, `max` (they show `NaN`).



* **`std` (Standard Deviation)**: Tells you how spread out the data is. `marks` has a `std` of **17.75**, meaning most students' marks are within ±17.75 of the average.
